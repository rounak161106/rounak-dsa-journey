```python
def binarySearch(l, ele):
    if not l:
        return "List is empty"
    low = 0
    high = len(l)-1
    while(low < high):
        mid = (low + high) // 2
        if l[mid] > ele:
            high = mid-1
        elif l[mid] < ele:
            low = mid+1
        elif l[mid] == ele:
            return f"Found at index {mid}"
    return "Not found"

print(binarySearch([3], 5))
```
