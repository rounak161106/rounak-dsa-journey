### A tree is a acyclic and minimally connected graph and has n-1 vertices. The Minimum Spanning Tree (MST) is defined only for connected, weighted, undirected graphs.

### 1. Prims Algorithm

```python
def prims_algo(ad_list):
    visited, distance, edges = {}, {}, [] #distance is just used to store the minimum cost needed to join a vertex to the MST.
    for i in ad_list:
        visited[i], distance[i] = False, float('inf')

    visited[0] = True
    for v,d in ad_list[0]:
        distance[v] = d

    for i in range(len(ad_list.keys())-1):
        mini = float('inf')
        next_vertex = None
        for u in ad_list:
            for v,d in ad_list[u]:
                if visited[u] and not visited[v] and d < mini:
                    next_vertex, mini, next_edge = v, d, (u,v)

        visited[next_vertex] = True
        edges.append(next_edge)
        for k,d in ad_list[next_vertex]:
            if not visited[k]:
                distance[k] = min(distance[k], d)
    return edges, distance

def prims_algo2(ad_list):
    visited, distance, nbr = {}, {}, {}
    for i in ad_list:
        visited[i], distance[i], nbr[i] = False, float('inf'), -1

    visited[0] = True
    for v,d in ad_list[0]:
        distance[v] = d
        nbr[v] = 0

    for i in range(len(ad_list.keys()) -1):
        minw = min([distance[i] for i in ad_list if not visited[i]])
        vertices = [v for v in ad_list if not visited[v] and distance[v] == minw]
        v = min(vertices)

        visited[v] = True
        for k, d in ad_list[v]:
            if not visited[k]:
                distance[k], nbr[k] = min(distance[k], d), v

    return nbr, distance

ad_list = {
    0: [(1, 4), (2, 3)],
    1: [(0, 4), (2, 1), (3, 2)],
    2: [(0, 3), (1, 1), (3, 4), (4, 5)],
    3: [(1, 2), (2, 4), (4, 2)],
    4: [(2, 5), (3, 2)]
}
edges, distance = prims_algo(ad_list)
print(edges)
print(distance)
nbr, distance = prims_algo2(ad_list)
print(nbr)
print(distance)
```

## Kruskal's Algorithm

```python
def kruskals(ad_list):
    edges, components, tree_edges = [], {}, []
    for u in ad_list:
        edges.extend([(d,u,v) for v,d in ad_list[u]])
        components[u] = u

    edges.sort()

    for d,u,v in edges:
        if components[u] != components[v]:
            tree_edges.append((u,v))
            c = components[u]
            for i in components:
                if components[i] == c:
                    components[i] = components[v]
    return tree_edges

ad_list = {
    0: [(1, 4), (2, 3)],
    1: [(0, 4), (2, 1), (3, 2)],
    2: [(0, 3), (1, 1), (3, 4), (4, 5)],
    3: [(1, 2), (2, 4), (4, 2)],
    4: [(2, 5), (3, 2)]
}
print(kruskals(ad_list))
```
