## Priority Queues
We have the concept of priority queues which supports two major operations, popping the element with the highest priority and pushing an element with a priority. Its usage is such that if we pop an element, we get the one with the highesh priority. One way to implement this is by using heap data structure.

## Heaps
- They are filled level by level from left to right.
- In max-heap, the value at each node is at least as big the value of its children and the root has the largest value.
