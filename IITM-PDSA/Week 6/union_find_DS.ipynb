## The complexity of Kruskal's algorithm was O(n2). We need to improve this and so we have union find data structure

## Union Find Data Structure

```python
class UnionFind:
    def __init__(self):
        self.components = {}
        self.members = {}
        self.size = {}

    def make_union_find(self, vertices):
        for i in vertices:
            self.components[i] = i
            self.members[i] = [i]
            self.size[i] = 1

    def find(self, vertex):
        return self.components[vertex]

    def union(self, u, v):                    #The complexity of one union operation is O(log n)    
        c1 = self.components[u]
        c2 = self.components[v]
        if self.size[c1] <= self.size[c2]:
            for i in  self.members[c1]:
                self.components[i] = c2
                self.members[c2].append(i)
            self.size[c2] += self.size[c1]
            self.members[c1] = []
        else:
            for i in  self.members[c2]:
                self.components[i] = c1
                self.members[c1].append(i)
            self.size[c1] += 1
            self.members[c2] = []
```

```python
def improved_kruskals(edges, vertices):
    edges = [(d,u,v) for u,v,d in edges]
    edges.sort()

    mf = UnionFind()
    mf.make_union_find(vertices)
    tree_edges = []
    
    for d,u,v in edges:
        if mf.find(u) != mf.find(v):
            mf.union(u,v)
            tree_edges.append((u,v,d))

    return tree_edges

edges = [(0,1,10),(0,2,18),(0,3,6),(0,4,20),(0,5,13),(1,2,10),(1,3,10),(1,4,5),(1,5,7),(2,3,2),(2,4,14),(2,5,15),(3,4,17),(3,5,12),(4,5,10)]
vertices = [0,1,2,3,4,5]
print(improved_kruskals(edges, vertices))
```

Therefore, the complexity reduces to O(m log n[for sorting] + n log n[for union]) = O((m+n)log n) overall
